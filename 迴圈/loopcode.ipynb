# loop 

loop指令可以讓我的程式碼依我的條件不斷執行，直到條件滿足為止



```python
#底下的程式碼會印出0-9
for i in range(10):
    print(i-1)
```

```python
list(range(10))
```

```python
for i in range(10):
    print(i-1)
```

```python
sum=0
for i in range(10):
    sum=sum+(i+1)

print(sum)    
```

```python
print(list(range(5)))
print(list(range(0,5)))
print(list(range(1,5)))
print(list(range(1,6,2)))
print(list(range(2,101,2)))
```
